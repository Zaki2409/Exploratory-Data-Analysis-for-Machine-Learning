```python
import pandas as pd

import matplotlib.pylab as plb
import matplotlib.pyplot as plt
%matplotlib inline

import seaborn as sns
```

```python
flora = pd.read_csv("Data/IRIS.csv")
flora.head(5)
```

```python
flora.groupby('species').mean().plot(color=['red', 'blue', 'black', 'green'],
                fontsize = 10.0,
                figsize=(4,4)
            )
```

```python
#seaborn plot, feature correlations
sns.pairplot(flora, hue='species', size=3)
```

```python
sns.jointplot(x= flora['sepal_length'], y= flora['sepal_width'], kind='hex')
```

```python
#Seaborn Plot, Facet Grid
#1st plot statment
plot= sns.FacetGrid(flora, col='species', margin_titles = True)
plot.map(plt.hist, 'sepal_width', color='green')
#2nd plot satement
plot = sns.FacetGrid(flora, col='species', margin_titles =True)
plot.map(plt.hist, 'sepal_length', color='blue')
```
