```python
import pandas as pd
import numpy as np

import matplotlib.pylab as plb
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
data = pd.read_csv("Data/IRIS.csv")
data.head(3)
```

```python
plt.plot(data.sepal_length, data.sepal_width, ls=' ', marker='o')
```

```python
#FIRST PLOT STATEMENT
plt.plot(data.sepal_length,
         data.sepal_width, ls=' ', marker='o',
         label='sepal')
#SECOND PLOT STATEMENT
plt.plot(data.petal_length,
         data.petal_width, ls=' ',marker='o',
         label='sepal')
```

```python
#Histogram
plt.hist(data.sepal_length, bins= 40)
```

```python
#Customizing plots
fig, ax = plt.subplots()
ax.barh(np.arange(10),
        data.sepal_width.iloc[:10])

#SET POSITIONS OF TICKS AND TICK LABEL
ax.set_yticks(np.arange(0.4, 10.4,1.0))
ax.set_yticklabels(np.arange(1,11))
ax.set(xlabel='xlabel', ylabel='ylabel',
       title='Title')
```
